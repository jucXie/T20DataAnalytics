Match Summary

```python
import pandas as pd
import json
```

```python
with open('E:\\BSc.IT\\TY\\DS Project\\t20_json_files\\t20_wc_match_results.json') as f:
    data = json.load(f)
#data[0]["matchSummary"]
len(data)
```

```python
df_match = pd.DataFrame(data[0]['matchSummary'])
df_match.head()
```

```python
df_match.shape
```

```python
df_match.rename({'scorecard': 'match_id'}, axis = 1, inplace = True)
df_match.head()
```

Create a match ids dictionary that maps team names to a unique match id. This will be useful later on to link with other tables

```python
match_ids_dict = {}
for index, row in df_match.iterrows():
    key1 = row['team1'] + " Vs " + row['team2']
    key2 = row['team2'] + " Vs " + row['team1']
    match_ids_dict[key1] = row['match_id']
    match_ids_dict[key2] = row['match_id']
len(match_ids_dict)  
```

```python
df_match.dtypes
```

```python
df_match['matchDate'] = pd.to_datetime(df_match['matchDate'])
df_match.head()
```

```python
specific_date = pd.to_datetime('2022-10-22')
df_match['stage'] = df_match['matchDate'].apply(lambda x: "Qualifier" if x < specific_date else "Super12")
df_match.tail(10)
```

```python
df=df_match[df_match["stage"]=='Qualifier'].count()
df
```

```python
df_match.dtypes
```

```python
df_match.to_csv('cricket_csv_files/match_summary.csv',index=False)
```

Batting Summary

```python
with open('E:\\BSc.IT\\TY\\DS Project\\t20_json_files\\t20_json_files\\t20_wc_batting_summary.json') as f:
    data = json.load(f) 
    #print(len(data))
    #data[0]["battingSummary"]
    all_records = []
    for rec in data:
        all_records.extend(rec["battingSummary"])

df_batting = pd.DataFrame(all_records)
df_batting.head(11)
#len(df_batting)
```

```python
df_batting["out/not_out"] = df_batting['dismissal'].apply(lambda x: "out" if len(x)>0 else "not_out")
df_batting.head(11)
```

```python
df_batting.drop('dismissal',axis=1,inplace=True)
```

```python
df_batting.head(11)
```

```python
df_batting['match_id'] = df_batting['match'].map(match_ids_dict) 
df_batting.loc[-5:]
```

Cleanup strings by removing special characters from names

```python
df_batting['batsmanName'] = df_batting['batsmanName'].apply(lambda x: x.replace('â€', ''))
df_batting['batsmanName'] = df_batting['batsmanName'].apply(lambda x: x.replace('\xa0', ''))
df_batting['batsmanName'] = df_batting['batsmanName'].apply(lambda x: x.replace('(c)', ''))

df_batting.head()
```

Renaming column "out/not_out" to "out"
out = 1
not_out = 0

```python
df_batting["out"] = df_batting['out/not_out'].apply(lambda x: 1 if x=="out" else 0)
df_batting.head(11)
```

```python
df_batting.drop('out/not_out',axis=1,inplace=True)
df_batting.head()
```

Changing the column name '4s' to 'fours' and the column name '6s' to 'sixes'.

```python
df_batting.rename(columns={'4s':'fours','6s':'sixes'},inplace=True)
```

```python
df_batting.head()
```

```python
df_batting.dtypes
```

```python
df_batting.to_csv('cricket_csv_files/batting_summary.csv',index=False)
```
